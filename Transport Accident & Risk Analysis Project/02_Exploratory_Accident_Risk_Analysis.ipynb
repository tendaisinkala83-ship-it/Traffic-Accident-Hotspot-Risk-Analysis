```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

```python
df= pd.read_csv('cleaned_accidents.csv')
df.head()
```

```python
temp_counts= df['Temperature(F)'].value_counts().sort_index()
```

```python
plt.figure()
plt.bar(temp_counts.index, temp_counts.values)
plt.title('Accident Temperature Distribution')
plt.xlabel('Temperature Level')
plt.ylabel('Number of Accidents')
plt.show()
```

Most accidents occur during moderate temperature conditions rather than extreme cold or heat. this may indicate that higher traffic activity during normal weather contributes more to accident frequency than temperature extremes.

```python
hourly_accidents= df['Hour'].value_counts().sort_index()
```

```python
plt.figure()
plt.plot(hourly_accidents.index, hourly_accidents.values)
plt.title('Accidents by Hour')
plt.xlabel('Hour')
plt.ylabel('Number of Accidents')
plt.show()
```

Accident frequency peaks during commuting hours, suggesting traffic congestion and increased road usage significantly influence accident occurence.

```python
day_order=[
    'Monday', 'Tuesday', 'Wednesday', 
    'Thursday', 'Friday', 'Saturday','Sunday'
]
daily_accidents= df['Day_of_Week'].value_counts().reindex(day_order)
```

```python
plt.figure()
plt.bar(daily_accidents.index, daily_accidents.values)
plt.title('Accidents by Day of Week')
plt.xticks(rotation=45)
plt.ylabel('Number of Accidents')
plt.show()
```

Weekday accident volumes exceed weekend levels, reflecting commuting-related transportation risk patterns.

```python
top_states= df['State'].value_counts().head(10)
```

```python
plt.figure()
plt.bar(top_states.index, top_states.values)
plt.title('Top 10 States by Accident Frequency')
plt.xticks(rotation=45)
plt.ylabel('Accident Count')
plt.show()
```

Accident concentration varies significantly by state, likely reflecting differences in population density, traffic volume, and infrastructure complexity.
