```python
import pandas as pd
import folium
from folium.plugins import HeatMap
```

```python
df= pd.read_csv('Cleaned_accidents.csv')
df.head()
```

```python
map_center = [df['Start_Lat'].mean(), df['Start_Lng'].mean()]

accident_map = folium.Map(location=map_center, zoom_start= 4)

accident_map
```

```python
sample_df = df.sample(n=2000, random_state= 42)
```

```python
for _, row in sample_df.iterrows():
    folium.CircleMarker(
        location= [row['Start_Lat'], row['Start_Lng']],
        radius=2,
        fill=True
    ).add_to(accident_map)

accident_map
```

```python
heat_data = df[['Start_Lat', 'Start_Lng']].values.tolist()
```

```python
heat_map = folium.Map(location=map_center, zoom_start=4)

HeatMap(heat_data[:10000]).add_to(heat_map)

heat_map
```

```python
heat_map.save('accident_heatmap.html')
```

Geospatial visualization reveals strong clustering of traffic accidents in high density metropolitan regions. Accident hotspots align with major urban corridors and interstate systems, suggesting traffic volume and infrstructure complexity are key contribution factors to risk concentration.
