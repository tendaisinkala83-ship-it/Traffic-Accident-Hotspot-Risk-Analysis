```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
df= pd.read_csv(r'/Users/tendaisinkala/Documents/Data Sets/USA_ACCIDENTS.csv')
```

```python
df.shape
df.head()
```

```python
df= df[[
    'ID',
    'Start_Time',
    'Start_Lat',
    'Start_Lng',
    'City',
    'State',
    'Temperature(F)'
]]
```

```python
df.info()
```

```python
df= df.dropna(subset=['Start_Lat', 'Start_Lng'])
```

```python
df= df.sample(n=100000, random_state=42)
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.to_csv('Cleaned_accidents.csv', index=False)
```

```python
df['Start_Time']= pd.to_datetime(
    df['Start_Time'],
    errors='coerce'
)
```

```python
df= df.dropna(subset=['Start_Time'])
```

```python
df['Year']= df['Start_Time'].dt.year
df['Month']=df['Start_Time'].dt.month
df['Hour']=df['Start_Time'].dt.hour
df['Day_of_Week']=df['Start_Time'].dt.day_name()
```

The dataset was reduced and cleaned to retain essential geographic, temporal, and environmental variables relevant for accident risk analysis. Records lacking geographic coordinates were removed to ensure accurate spatial visualization and mapping.
